## 1. Index Calculations (Pure Math)

```python
from housing_germany.indices.core import rebase_series, evolution_between

dates = pd.date_range('2020-01-01', periods=12, freq='MS')
values = [100, 102, 105, 103, 108, 110, 112, 115, 118, 120, 122, 125]
series = rebase_series(dates, values)
print('Rebased series:', series.head().tolist())

evol = evolution_between(series, '2020-01-01', '2020-12-01')
print(f'Evolution Jan-Dec 2020: {evol:.1%}')
```

## 2. Price Calculations

```python
from housing_germany.prices.core import implied_price_at

past_price = implied_price_at(3000, 0.15)
print(f'If current price is 3000 EUR/sqm and evolution was +15%,')
print(f'past price was: {past_price:.0f} EUR/sqm')
```

## 3. Purchasing Power Tools

```python
from housing_germany.purchasing_power.mortgage import mortgage_formula
from housing_germany.purchasing_power.tax import get_tax_rates_DE

factor = mortgage_formula(3.5, years=30)
print(f'Mortgage factor (PVIFA) at 3.5%: {factor:.2f}')
print(f'Monthly payment for 300k EUR loan: {300_000/factor:.0f} EUR')

print('\nGrunderwerbsteuer by Bundesland:')
for state, rate in sorted(get_tax_rates_DE().items()):
    print(f'  {state}: {rate:.1%}')
```
